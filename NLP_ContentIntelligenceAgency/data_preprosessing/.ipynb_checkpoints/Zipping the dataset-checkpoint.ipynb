```python
import pandas as pd
import zipfile
transcript_2 = "transformers_dataset.csv"
df = pd.read_csv(transcript_2)
# Save the dataset as a CSV file
df.to_csv("transformers_dataset.csv", index=False, encoding="utf-8")

# Create a ZIP file and add the CSV file
with zipfile.ZipFile("transformers_dataset.zip", "w", zipfile.ZIP_DEFLATED) as zipf:
    zipf.write("transformers_dataset.csv")
```
